# Dataset Exploration
This notebook performs dataset exploration as required by Phase 2. 
It calculates image counts, dimensions, intensity distributions, and visualizes overlays.

```python
import os
import glob
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from pathlib import Path
from src.utils.config import load_config

# Ensure outputs directory exists
os.makedirs('../outputs/figures', exist_ok=True)

config = load_config('../config.yaml')
seg_dir = config.path('segmentation_data')
print(f"Exploring dataset at: {seg_dir}")
```

```python
extensions = config.values['data']['supported_image_extensions']
all_files = []
for ext in extensions:
    all_files.extend(list(seg_dir.rglob(f'*{ext}')))

images = [f for f in all_files if 'mask' not in f.name.lower()]
masks = [f for f in all_files if 'mask' in f.name.lower()]

print(f"Total Images: {len(images)}")
print(f"Total Masks: {len(masks)}")
```

```python
if images and masks:
    # Sample visualization
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    img = Image.open(images[0]).convert('L')
    mask = Image.open(masks[0]).convert('L')
    
    axes[0].imshow(img, cmap='gray')
    axes[0].set_title('Original Image')
    axes[0].axis('off')
    
    axes[1].imshow(mask, cmap='gray')
    axes[1].set_title('Segmentation Mask')
    axes[1].axis('off')
    
    axes[2].imshow(img, cmap='gray')
    axes[2].imshow(mask, cmap='jet', alpha=0.3)
    axes[2].set_title('Overlay')
    axes[2].axis('off')
    
    plt.savefig('../outputs/figures/image_mask_overlay.png')
    plt.show()
```
