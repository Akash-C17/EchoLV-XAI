# ROI Extraction (Phase 4)
This notebook visualizes the left ventricle (LV) region-of-interest extraction process.

```python
import os
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
from src.preprocessing.roi_processing import extract_roi, apply_mask

os.makedirs('../outputs/figures', exist_ok=True)

# Synthesize an example for demonstration
img = Image.new('L', (256, 256), color=100)
mask = Image.new('L', (256, 256), color=0)
mask_arr = np.array(mask)
mask_arr[100:150, 120:180] = 1
mask = Image.fromarray(mask_arr)

cropped_roi = extract_roi(img, mask, padding=15)
masked_img = apply_mask(img, mask)
masked_roi = extract_roi(masked_img, mask, padding=15)

fig, axes = plt.subplots(1, 4, figsize=(20, 5))
axes[0].imshow(img, cmap='gray'); axes[0].set_title('Original')
axes[1].imshow(mask, cmap='gray'); axes[1].set_title('Mask')
axes[2].imshow(cropped_roi, cmap='gray'); axes[2].set_title('Cropped LV ROI')
axes[3].imshow(masked_roi, cmap='gray'); axes[3].set_title('Masked LV ROI')

for ax in axes: ax.axis('off')
plt.savefig('../outputs/figures/roi_extraction_example.png')
plt.show()
```
